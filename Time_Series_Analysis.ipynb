```python
%%R
library(forecast)

# Automatically select best ARIMA model
arima_model <- auto.arima(AirPassengers)

# Model summary
summary(arima_model)
```

```python
%%R
# Forecast next 12 months using ARIMA
arima_forecast <- forecast(arima_model, h = 12)

# Plot forecast
plot(arima_forecast,
     main = "ARIMA Forecast for AirPassengers")
```

```python
%%R
accuracy(arima_forecast)
```

```python
%%R
library(forecast)

# Auto SARIMA model
sarima_model <- auto.arima(
  AirPassengers,
  seasonal = TRUE
)

# Model summary
summary(sarima_model)
```

```python
%%R
sarima_forecast <- forecast(sarima_model, h = 12)

plot(sarima_forecast,
     main = "SARIMA Forecast for AirPassengers")
```

```python
%%R
accuracy(sarima_forecast)
```

```python
%%R
manual_sarima <- Arima(
  AirPassengers,
  order = c(0,1,1),
  seasonal = list(order = c(0,1,1), period = 12)
)

summary(manual_sarima)

manual_forecast <- forecast(manual_sarima, h = 12)
plot(manual_forecast)
```

📌 Residual Diagnostics for SARIMA Model


Goal: Check whether residuals behave like white noise
(i.e., no pattern, zero mean, constant variance, no autocorrelation)

```python
%%R
# Extract residuals
residuals_sarima <- residuals(sarima_model)

# Summary
summary(residuals_sarima)
```

```python
%%R
plot(residuals_sarima,
     main = "Residuals of SARIMA Model",
     ylab = "Residuals",
     xlab = "Time")
abline(h = 0, col = "red")
```

```python
%%R
hist(residuals_sarima,
     main = "Histogram of SARIMA Residuals",
     col = "lightblue",
     xlab = "Residuals")
```

```python
%%R
qqnorm(residuals_sarima)
qqline(residuals_sarima, col = "red")
```

```python
%%R
hist(residuals_sarima,
     main = "Histogram of SARIMA Residuals",
     col = "lightblue",
     xlab = "Residuals")
```

```python
%%R
qqnorm(residuals_sarima)
qqline(residuals_sarima, col = "red")
```

```python
%%R
acf(residuals_sarima,
    main = "ACF of SARIMA Residuals")
```

```python
%%R
Box.test(residuals_sarima,
         lag = 20,
         type = "Ljung-Box")
```

✅ When Model is GOOD

✔ Residual mean ≈ 0
✔ No trend/seasonality
✔ ACF spikes inside limits
✔ Ljung-Box p-value > 0.05

❌ When Model is BAD

❌ Patterns in residual plot
❌ Significant ACF spikes
❌ Ljung-Box p-value < 0.05



🔹 Why Heteroscedasticity is a Problem

Makes confidence intervals unreliable

Affects accuracy of forecasts

Violates model assumptions

Standard errors become incorrect

🔹 In Time Series (ARIMA / SARIMA)

In time series, heteroscedasticity means:

Volatility changes over time

Residual variance increases/decreases

Common in financial data (stocks, returns)

🔹 How to Detect Heteroscedasticity
1️⃣ Residual Plot

Fan-shaped pattern → heteroscedasticity

2️⃣ ARCH Test (Most Common)
ArchTest(residuals_sarima)


p-value < 0.05 → heteroscedasticity present ❌

p-value ≥ 0.05 → no heteroscedasticity ✅

🔹 How to Handle Heteroscedasticity

Log transformation

Box-Cox transformation

Use GARCH models

Robust standard errors
